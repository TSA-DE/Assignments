1. Create a 3x3 array with random integers between 1 and 100. Write a program to "normalize" the array so that the values are scaled between 0 and 1. • Formula: (x - min) / (max - min) • Example Output: An array where the smallest value becomes 0.0 and the largest becomes 1.0.

````python
import numpy as np
````

````python
from numpy import random 
````

````python
arr=random.randint(1,100,size=(3,3))
````

````python
print(arr)
````

````python
arr_normalized=(arr-arr.min())/(arr.max()-arr.min())
print(arr_normalized)
````

Create a 6x4 matrix of random integers. Identify which rows have a sum that is an even number. Print the indices of these rows.

````python
arr_4=random.randint(1,100,size=(6,4))
arr_4
````

````python
sum=0
for i in arr_4 :
    print(i)
    for j in i :
        sum=sum+j
        #print(j)
    
    if sum %2==0:
        print("sum is even and = : ",sum)
        idx=np.where((arr_4==i).all(axis=1))[0]
        print("row_number is : ",idx)
        print("===============================")
    else : 
        
        print("sum is odd and = : ",sum)
        print("===============================")


    



        
````
