1. Ask the user for a string that contains both letters and numbers (e.g., "Py2th0on3"). Write a program that separates them into two lists: one for the letters and one for the numbers. Calculate the sum of the numbers found and join the letters to form a word. • Example Input: "Py2th0on3" • Example Output: Word: "Python", Sum: 5

```python
str1=input("please enter a string : ")
lst1=list(str1)
#print(lst1)
lst2=[]
lst3=[]
sum=0
for e in lst1:
    if e.isalpha()==True:
        lst2.append(e)
    elif e.isdigit()==True:
        lst3.append(e)
print("Word: ","".join(lst2))
for i in lst3:
    sum=sum+int(i)
#print(lst3)
print ("sum : ",sum)

```

 8. Given a list of numbers with many duplicates, return a list of tuples showing each unique element
 and its frequency (how many times it appeared).
 • Example Input: [1, 1, 2, 3, 3, 3, 4]
 • Example Output: [(1, 2), (2, 1), (3, 3), (4, 1)]


```python
numbers = [1, 1, 2, 3, 3, 3, 4]

counts = {}

for num in numbers:
    counts[num] = counts.get(num, 0) + 1

result = list(counts.items())

print(result)
    

```

 10.Start with a dictionary of items and prices: {'apple': 0.5, 'bread': 2.0, 'milk': 1.5}. Allow the user to
 "Add" or "Remove" items from a cart. If they add an item not in the store, print an error. Finally, print
the total bill.

```python
store={"apple":0.5,"bread":2.0,"milk":1.5}
cart=[]
action=input("please choose an action (add,remove)").lower()
if action =="add":
    item=input("please enter the item you would like to add : ").lower()
    if item in store.keys():
        cart.append(item)
        print(item,"is added to the cart")
    else :
        print("item is not available at the store")
elif action=="remove":
    item=input("please enter the item you would likee to remove : ").lower()
    if item in cart :
        cart.remove(item)
        print(item,"is removed from the cart")
    else :
        print("error")
#calculating total bill
sum=0
for item in cart:
    sum=sum+store[item]
print("total bill is : ",sum)



```

11.Check if a full sentence is a palindrome, ignoring spaces and capitalization.
• Example Input: "Race car"
• Example Output: True

```python
word=list(input("please enter a word: ").replace(" ","").lower())
print(word)
reversed_word=[]
for l in reversed(word):
    reversed_word.append(l)
print(reversed_word)
if word == reversed_word:
    print("scentence is a palindrome")
else :
    print("scentence is not a palindrome")
```

14.Take a dictionary like {'A': 1, 'B': 2, 'C': 1} and flip it. If two keys have the same value, the new value should be a list of those keys. • Example Input: {'A': 1, 'B': 2, 'C': 1} • Example Output: {1: ['A', 'C'], 2: ['B']}

```python
dict1={"A":1,"B":2,"C":1}
dict2={}
for key,value in dict1.items():
    dict2.setdefault(value,[]).append(key)
        
print(dict2)
```

15. Create a dictionary where the keys are the vowels found in a user's string and the values are lists of the indices where those vowels appear. • Example Input: "banana" • Example Output: {'a': [1, 3, 5]}

```python
w=list(input("please enter a word with vowels : ").lower())
#print(w)
list1=[]
dict1={}
for i in range(int(len(w))) :
    if w[i] in ["a","e","i","o","u"]:
        list1.append(i)

        dict1.setdefault(w[i],list1)
#print(list1)
print(dict1)
```
